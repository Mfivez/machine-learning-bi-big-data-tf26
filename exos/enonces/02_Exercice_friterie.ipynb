# Jour 02 · Classification binaire à la friterie

Au début de chaque quart d’heure, tu connais le nombre de commandes en file. À partir de ce seul nombre, prédis la classe du prochain client : **0** si son attente sera de 15 minutes ou moins, **1** si elle dépassera 15 minutes.

Le fichier [friterie.csv](../../data/friterie.csv) rassemble 60 services simulés sans renfort. Chaque ligne correspond à un créneau. La colonne `attente_longue` indique la classe observée après le passage du client.

**1.** Charge le CSV et vérifie les types, les valeurs manquantes, les doublons et les valeurs prises par la cible.

**2.** Prépare la variable d’entrée `commandes_en_attente` et la cible `attente_longue`. Garde aussi `date_service` pour identifier le dernier service et `heure` pour y simuler les appels dans l’ordre à l’étape 8.

**3.** Réserve le dernier service entier pour la mise en situation finale. Sur les créneaux des autres services, reproduis le partage stratifié du cours entre apprentissage (60 %), validation (20 %) et test (20 %). Garde le test de côté jusqu’au bilan final.

**4.** Sur l’entraînement, mesure la fréquence des longues attentes et représente leur lien avec le nombre de commandes en attente.

**5.** Entraîne deux classifieurs sur les mêmes données : une régression logistique et un arbre de décision. Compare les probabilités qu’ils attribuent à la classe 1 pour plusieurs longueurs de file.

**6.** Sur la validation, compare deux modèles à trois seuils. Relève les alertes, les longues attentes retrouvées, la précision et le rappel. Avec un budget global de trois places sur douze, choisis la règle qui retrouve le plus de longues attentes. À résultat égal, préfère la liste la plus courte. Compare avec quatre places sur douze.

**7.** Sur le test, compare les classes prédites aux classes observées. Construis la matrice de confusion, calcule l’exactitude, la précision, le rappel et le score F1. Explique ce que représentent un faux positif et un faux négatif à la friterie.

**8.** Applique la règle à trois renforts au dernier service réservé au départ. Appelle sur les créneaux prédits en classe 1, dans l’ordre, jusqu’à trois appels. Combien de longues attentes auraient donné lieu à un appel ? Combien seraient restées sans appel ?
